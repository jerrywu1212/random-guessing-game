```python
import random
```

```python
answer = random.randint(1, 10000000)
guess = random.randint(1, 10000000)

while answer == guess:
    guess = random.randint(1, 10000000)

def testWhileLoop(guess, answer):
    number_of_iterations = 0
    while guess != answer:
        number_of_iterations += 1
        guess = random.randint(1, 10000000)
    print("It took " + str(number_of_iterations) + " iterations to guess the right answer!")

testWhileLoop(guess, answer)
    

    
```

```python
import time

start = time.time()

total = 0
for i in range(10_000_000):
    total += i

end = time.time()
print(f"Took {end - start:.4f} seconds")
```
